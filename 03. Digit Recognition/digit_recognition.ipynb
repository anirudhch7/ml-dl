1. Data Loading and Initial Inspection

The MNIST dataset is loaded, containing 60,000 training images and 10,000 test images, each of size $28 \times 28$ pixels.

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt

# Load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Inspect shape and visualize a sample
print(x_train.shape) # Output: (60000, 28, 28)
plt.imshow(x_train[0], cmap=plt.cm.binary)
plt.show()
```

2. Preprocessing and Reshaping

The pixel values are normalized to a range between 0 and 1 to improve model convergence. Additionally, the data is reshaped to include a color channel (1 for grayscale), which is required for CNN layers.

```python
import numpy as np

# Normalizing pixel values
x_train = tf.keras.utils.normalize(x_train, axis=1)
x_test = tf.keras.utils.normalize(x_test, axis=1)

# Reshaping for CNN (adding the channel dimension)
IMG_SIZE = 28
x_trainr = np.array(x_train).reshape(-1, IMG_SIZE, IMG_SIZE, 1)
x_testr = np.array(x_test).reshape(-1, IMG_SIZE, IMG_SIZE, 1)

print("Training dimension:", x_trainr.shape) # (60000, 28, 28, 1)
```

3. Building the CNN Architecture

The model uses a Sequential stack consisting of three convolutional layers (each with 64 filters), ReLU activation, and Max Pooling to extract features, followed by dense layers for final classification into 10 categories (digits 0-9).

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Conv2D, MaxPooling2D, Activation

model = Sequential()

# Layer 1
model.add(Conv2D(64, (3,3), input_shape=x_trainr.shape[1:]))
model.add(Activation("relu"))
model.add(MaxPooling2D(pool_size=(2,2)))

# Layer 2 & 3
for _ in range(2):
    model.add(Conv2D(64, (3,3)))
    model.add(Activation("relu"))
    model.add(MaxPooling2D(pool_size=(2,2)))

# Classification Layers
model.add(Flatten())
model.add(Dense(64, activation="relu"))
model.add(Dense(32, activation="relu"))
model.add(Dense(10, activation="softmax"))

model.summary()
```

4. Training and Evaluation

The model is compiled with the Adam optimizer and Sparse Categorical Crossentropy loss. It is trained for 5 epochs, achieving a validation accuracy of approximately 91.2%.

```python
# Compilation
model.compile(loss="sparse_categorical_crossentropy", 
optimizer="adam", 
metrics=['accuracy'])

# Training
model.fit(x_trainr, y_train, epochs=5, validation_split=0.3)

# Evaluation
test_loss, test_acc = model.evaluate(x_testr, y_test)
print("Validation Accuracy:", test_acc)
```

5. Prediction and Testing

The model makes predictions on the test set. Using np.argmax, the class with the highest probability is selected as the final digit prediction.

```python
predictions = model.predict([x_testr])

# Predicting the first test sample
print("Predicted digit:", np.argmax(predictions[0]))

# Visualizing the sample to verify
plt.imshow(x_test[0])
plt.show()
```
