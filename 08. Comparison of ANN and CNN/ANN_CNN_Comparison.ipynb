# MNIST Classification: ANN vs CNN Comparison

This code compares the performance and training efficiency of two popular neural network architectures using the MNIST dataset of handwritten digits:

1.  **Artificial Neural Network (ANN):** A basic feed-forward architecture.
2.  **Convolutional Neural Network (CNN):** A specialized architecture for image data that uses spatial features.

## 1. Load and Preprocess the Dataset
We begin by importing the necessary libraries, loading the data, and normalizing it.

```python
import tensorflow as tf
import time

# Load and Preprocess the Dataset
print("Loading MNIST dataset...")
mnist = tf.keras.datasets.mnist
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# Normalize pixel values to a range of 0 to 1
train_images = train_images / 255.0
test_images = test_images / 255.0

# Reshape data for the CNN (Adding a single color channel for grayscale: 28x28x1)
train_images_cnn = train_images.reshape(-1, 28, 28, 1)
test_images_cnn = test_images.reshape(-1, 28, 28, 1)
```

## 2. Build and Train the ANN Model
The ANN flattens the input image and uses standard dense layers.

```python
print("\n--- Building and Training ANN ---")
ann_model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),          # Flatten to 1D (784)
    tf.keras.layers.Dense(128, activation='relu'),          # Hidden layer
    tf.keras.layers.Dropout(0.2),                           # Regularization
    tf.keras.layers.Dense(10, activation='softmax')         # Output layer (10 classes)
])

ann_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

# Time the training process
start_time = time.time()
ann_model.fit(train_images, train_labels, epochs=5, validation_split=0.2, verbose=2)
ann_train_time = time.time() - start_time
```

## 3. Build and Train the CNN Model
The CNN uses convolutional and pooling layers to extract spatial features before reaching the classification layers.

```python
print("\n--- Building and Training CNN ---")
cnn_model = tf.keras.Sequential([
    # First Convolutional Block
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28, 28, 1)),
    tf.keras.layers.MaxPooling2D(2, 2),
    
    # Second Convolutional Block
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2, 2),
    
    # Flatten the 2D feature maps and pass to Dense layers
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

# Time the training process
start_time = time.time()
cnn_model.fit(train_images_cnn, train_labels, epochs=5, validation_split=0.2, verbose=2)
cnn_train_time = time.time() - start_time
```

## 4. Evaluate and Compare Results
Finally, we compare the accuracy and training time of both models.

```python
print("\n==========================================")
print("             FINAL COMPARISON               ")
print("==========================================")

# Evaluate ANN
ann_loss, ann_acc = ann_model.evaluate(test_images, test_labels, verbose=0)
print(f"ANN Test Accuracy:  {ann_acc * 100:.2f}%")
print(f"ANN Training Time:  {ann_train_time:.2f} seconds")

print("-" * 42)

# Evaluate CNN
cnn_loss, cnn_acc = cnn_model.evaluate(test_images_cnn, test_labels, verbose=0)
print(f"CNN Test Accuracy:  {cnn_acc * 100:.2f}%")
print(f"CNN Training Time:  {cnn_train_time:.2f} seconds")
print("==========================================")
```
