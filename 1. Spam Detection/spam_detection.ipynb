1. Environment Setup & Data Loading

The process begins by importing necessary libraries for data manipulation (pandas, numpy), machine learning (sklearn), and loading the dataset from a local CSV file.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Loading the data
df = pd.read_csv('C:\\Users\\anirudhch\\Desktop\\mail_data.csv')

# Handling null values by replacing them with empty strings
data = df.where((pd.notnull(df)), "")
```

2. Data Preprocessing & Label Encoding

The textual categories are converted into numerical values to make them compatible with the Logistic Regression model.

Spam is encoded as 0

Ham is encoded as 1

```python
# Label encoding
data.loc[data['Category'] == 'spam', 'Category'] = 0
data.loc[data['Category'] == 'ham', 'Category'] = 1

# Splitting features (X) and target (Y)
X = data['Message']
Y = data['Category']
```

3. Train-Test Split

The data is split into a training set ($80\%$) and a testing set ($20\%$) to evaluate the model's performance on unseen data.

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3)
```

4. Feature Extraction (TF-IDF)

Raw text is converted into numerical feature vectors using the TfidfVectorizer. This accounts for word frequency while penalizing overly common words (like "the" or "is").

```python
feature_extraction = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)

# Transform text to feature vectors
X_train_features = feature_extraction.fit_transform(X_train)
X_test_features = feature_extraction.transform(X_test)

# Ensure target labels are integers
Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

5. Model Training & Evaluation

A Logistic Regression model is initialized and trained on the extracted features. The code reports a training accuracy of approximately 96.7%.

```python
model = LogisticRegression()
model.fit(X_train_features, Y_train)

# Checking accuracy on training data
prediction_on_training_data = model.predict(X_train_features)
accuracy_on_training_data = accuracy_score(Y_train, prediction_on_training_data)
print('Acc on training data:', accuracy_on_training_data)
```

6. Predictive System

Finally, the model is used to predict whether a new, custom email string is spam or ham.

```python
input_your_mail = ["Dear Sir, get your lifetime free Credit Card. Now! "]

# Convert input to features
input_data_features = feature_extraction.transform(input_your_mail)

# Prediction
prediction = model.predict(input_data_features)

if (prediction[0] == 1):
    print('Ham mail')
else:
    print('Spam mail')
```
