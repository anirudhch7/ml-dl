# ANN Classification with Fashion MNIST

This code demonstrates a simple feed-forward Artificial Neural Network (ANN) using TensorFlow / Keras to classify the Fashion MNIST dataset. The notebook is divided into sections: data loading, model definition, compilation & training, evaluation, and a prediction example.

```python
# Imports and data loading
import tensorflow as tf
import numpy as np

print("Loading data...")
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Normalize pixel values to [0, 1]
train_images = train_images / 255.0
test_images = test_images / 255.0

# Class names for reference
class_names = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]
```

## Model Architecture

We use a small fully-connected network: a flatten input layer, one hidden Dense layer with ReLU, a Dropout for regularization, and a final Dense layer with Softmax for 10-class classification.

```python
# Build the ANN model
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),  # flatten 28x28 -> 784
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
])
```

## Compile and Train

We compile the model with the Adam optimizer and `sparse_categorical_crossentropy` loss (labels are integer encoded). Training uses a validation split so you can monitor generalization during training.

```python
# Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
print("\nStarting model training...")
history = model.fit(train_images, train_labels, epochs=10, validation_split=0.2)
```

## Evaluation and Prediction

Evaluate the model on the test set and run a single-sample prediction to compare predicted vs actual class.

```python
# Evaluate on test data
print("\nEvaluating model on test data...")
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f'\nFinal Test Accuracy: {test_acc:.4f}')

# Make predictions
print("\nMaking a prediction on the first test image...")
predictions = model.predict(test_images)
predicted_index = np.argmax(predictions[0])
actual_index = test_labels[0]
print(f"Predicted Class: {class_names[predicted_index]} (Confidence: {predictions[0][predicted_index]*100:.2f}%)")
print(f"Actual Class:    {class_names[actual_index]}")
```
