```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'C:\Users\CVR\Desktop\ML\groceries2.csv',header=None)
df.isnull().sum()
```

```python
T=[]
for i in range(len(df)):
  T.append([str(df.values[i,j]) for j in range(0,4) if str(df.values[i,j])!='nan'])
T
```

```python
from apyori import apriori
rules=apriori(T,min_support=0.03,min_confidence=0.35,min_lf=3,min_length=2)
```
